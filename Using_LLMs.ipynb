# Day 4: Using LLMs

There are many ways to test the capabilities of LLMs. You already use one of them: ChatGPT. In that sense, ChatGPT is the *playground*—the public interface—that lets you interact with a model without writing any code.

But you want to use LLMs not only to debug code or write essays or as search engine; you want to use LLM as a **researcher**. And for research purposes, you often want more control. You may want to decide exactly what prompt to send, which model to use, what parameters to change, and how to store the output. To do that, you usually access a model through an **API**.

To that end, you want to access the API directly. Now, OpenAI might say it is a non-profit organization, but its business model is very profit-like. Thus, accessing their API costs money, yet, not a lot if you are using a cheap model (e.g., GPT-4o mini) and just testing it out. Alternatively, other models like Mistral or Llama are open-source, so you do not need to pay to use them... However, you won't be able to run them locally, which means you will still need to pay a cloud service that has those models and you can deploy them remotely (e.g., [Deep Infra](https://deepinfra.com/)). The latter option is still cheaper.

Regardless, the code and structure is the same for all LLMs.

## Before you begin

A few practical notes:

- **API access usually costs money.** For simple examples, the cost is often tiny, but costs can add up in larger projects.
- **Different providers host different models.** The general workflow is often similar even when the model changes.
- **Results are not perfectly stable.** If you change the prompt, the model, or the temperature, you may get a different answer.
- **The code here is meant to teach the logic.** Later, we will build more systematic workflows for research tasks.


## 1. Accessing the OpenAI API

```python
# Core imports
from openai import OpenAI

# Create a client.
# Replace "YOUR KEY" with your real API key.
client = OpenAI(api_key="YOUR KEY")

# A prompt is a list of messages.
# The system message sets the general behavior of the model.
# The user message contains the actual request.
messages = [
    {
        "role": "system",
        "content": "You are a clear and concise research assistant."
    },
    {
        "role": "user",
        "content": (
            "In 3-4 sentences, explain why API access can be useful for "
            "computational social science research."
        )
    }
]

response = client.chat.completions.create(
    model="gpt-5-mini",
    messages=messages,
    temperature=0.3,   # lower = more consistent, higher = more varied
    max_tokens=150
)
```

```python
# Display the model's reply
print(response.choices[0].message.content.strip())
```

### What is happening here?

There are four key parts:

- **client**: authenticates you with the provider.
- **messages**: stores the conversation you want the model to see.
- **model**: specifies which model you want to call.
- **response**: stores the model's output.

This message-based structure is worth learning well because you will use it again and again in more advanced workflows.


## 2. Accessing other LLMs through DeepInfra

```python
from openai import OpenAI

# DeepInfra exposes an OpenAI-compatible API.
# That means the code looks very similar; the main differences are the base_url and model name.
client = OpenAI(
    base_url="https://api.deepinfra.com/v1/openai",
    api_key="YOUR KEY"
)

messages = [
    {
        "role": "system",
        "content": "You are a clear and concise research assistant."
    },
    {
        "role": "user",
        "content": (
            "In 3-4 sentences, explain why API access can be useful for "
            "computational social science research."
        )
    }
]

response = client.chat.completions.create(
    model="meta-llama/Llama-3.3-70B-Instruct",
    messages=messages,
    temperature=0.3,
    max_tokens=150
)
```

```python
# Display the model's reply
print(response.choices[0].message.content.strip())
```

## 3. What stays the same across providers?

Notice that most of the workflow does **not** change:

- you create a client,
- you write messages,
- you choose a model,
- you inspect the response.

What usually changes is:

- the **API key**,
- the **base URL**,
- the **model name**, and sometimes
- small provider-specific features or limits.

That is why learning the general structure is so useful.

## 4. A small improvement: use a helper function

Once you start experimenting, it is convenient to wrap the API call in a function. That makes your code easier to reuse and easier to debug.


```python
def get_completion(client, model, messages, temperature=0.3, max_tokens=150):
    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=temperature,
        max_tokens=max_tokens
    )
    return response.choices[0].message.content.strip()

# Example usage:
# answer = get_completion(client, "meta-llama/Llama-3.3-70B-Instruct", messages)
# print(answer)
```

## 5. Important limitations

This notebook shows the mechanics of calling an LLM, but not every response should be trusted.

Students should keep at least four limitations in mind:

- LLMs can produce inaccurate or fabricated information.
- Responses can change when the prompt changes.
- Providers may update models over time.
- Even cheap experiments can become expensive when scaled up.

So the main lesson is not just "how to get a response," but also "how to think critically about the response."

The main idea of this lesson is simple: once you understand the `client -> messages -> model -> response` pattern, you understand the backbone of many LLM workflows.

That same structure will later support more advanced tasks such as annotation, classification, summarization, extraction, and agentic pipelines.
